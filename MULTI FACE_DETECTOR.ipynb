```python
import cv2
import mediapipe as mp

# Start camera
cap = cv2.VideoCapture(0)

# MediaPipe Face Detection
mpFaceDetection = mp.solutions.face_detection
faceDetection = mpFaceDetection.FaceDetection(
    model_selection=0,
    min_detection_confidence=0.5
)

# Neon blue color
color = (255, 255, 0)

# Glow thickness controller
thickness = 2
direction = 1

while True:

    success, img = cap.read()

    if not success:
        break

    # Mirror effect
    img = cv2.flip(img, 1)

    # Convert to RGB
    imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Detect faces
    results = faceDetection.process(imgRGB)

    # Blinking glow effect
    thickness += direction

    if thickness >= 6:
        direction = -1

    if thickness <= 2:
        direction = 1

    # Draw all detected faces
    if results.detections:

        for detection in results.detections:

            bboxC = detection.location_data.relative_bounding_box

            h, w, c = img.shape

            x = int(bboxC.xmin * w)
            y = int(bboxC.ymin * h)
            width = int(bboxC.width * w)
            height = int(bboxC.height * h)

            # Draw glowing rectangle
            cv2.rectangle(
                img,
                (x, y),
                (x + width, y + height),
                color,
                thickness
            )

            # AI Label
            cv2.putText(
                img,
                "AI FACE DETECTED",
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.9,
                color,
                2
            )

    # Main title
    cv2.putText(
        img,
        "PPSL Smart AI Camera",
        (10, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        color,
        3
    )

    # Show screen
    cv2.imshow("AI MULTI FACE DETECTOR", img)

    # Quit with q
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release everything
cap.release()
cv2.destroyAllWindows()
```
